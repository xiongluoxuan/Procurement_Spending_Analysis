```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
purchase_date_text = df_clean["PurchaseDate"]
```

```python
purchase_date = pd.to_datetime(purchase_date_text)
```

```python
df_clean["PurchaseDate"] = purchase_date
```

```python
purchase_date_type = df_clean["PurchaseDate"].dtype
purchase_date_type
```

```python
supplier_text = df_clean["Supplier"]
supplier_text = supplier_text.str.strip()
```

```python
df_clean["Supplier"] = supplier_text
```

```python
category_text = df_clean["Category"]
category_text = category_text.str.strip()
```

```python
df_clean["Category"] = category_text
```

```python
spend_amount = df_clean["TotalCost"]
```

```python
df_clean["Spend"] = spend_amount
```

```python
month_value = purchase_date.dt.strftime("%Y-%m")
```

```python
df_clean["Month"] = month_value
```

```python
supplier_column = df_clean["Supplier"]
spend_column = df_clean["Spend"]
```

```python
supplier_groups = spend_column.groupby(supplier_column)
```

```python
supplier_spend = supplier_groups.sum()
supplier_spend
```

```python
supplier_spend_table = supplier_spend.reset_index()
supplier_spend_table
```

```python
supplier_spend_table = supplier_spend_table.sort_values("Spend", ascending=False)
supplier_spend_table
```

```python
category_column = df_clean["Category"]
category_groups = spend_column.groupby(category_column)
```

```python
category_spend = category_groups.sum()
category_spend
```

```python
category_spend_table = category_spend.reset_index()
category_spend_table
```

```python
category_spend_table = category_spend_table.sort_values("Spend", ascending=False)
category_spend_table
```

```python
month_column = df_clean["Month"]
month_groups = spend_column.groupby(month_column)
```

```python
monthly_spend = month_groups.sum()
monthly_spend
```

```python
monthly_spend_table = monthly_spend.reset_index()
monthly_spend_table
```
